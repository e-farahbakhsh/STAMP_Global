This notebook estimates the preservation-exposure score using cumulative erosion grids generated by landscape evolution models. For simplicity, this score will be referred to as the preservation score throughout the rest of the notebook. The results will refine the mineralisation probability calculated in notebook `05_prospectivity_map` and enable the calculation of preserved mineralisation prospectivity.

The workflow is as follows:
1. Coregister points with cumulative erosion grids.
2. Fit a case-control logistic regression, using a cubic B-spline basis in log-transformed cumulative erosion, that compares known mineral occurrences with unlabelled background points, and rescale the fitted response to a bounded preservation score. Comparing occurrences with the background recovers the relative enrichment of deposits at each level of erosion, rather than reproducing the erosion distribution of the arc area that was available to be sampled.
3. Multiply preservation score by mineralisation probability to produce a map of preserved mineralisation prospectivity.

This final map highlights areas where mineralisation is not only likely to have formed but also likely to have survived erosion and to sit at a depth where it could be detected. The preservation score is a bounded relative measure rather than a calibrated probability of survival, so the product should be read as an integrated formation-preservation index.

### Coregistration

This code block coregisters different sets of points generated in previous notebooks—including reconstructed locations of mineral occurrences, unlabelled samples, and target points within arc–back-arc environments and continental landmasses—with cumulative erosion grids.  

For each point, it assigns the mean value of the grid pixels within a user-defined square neighborhood. The side length of this square is controlled by the `distance` argument in the function below.

```python
columns_ero = ["present_lon", "present_lat", "lon", "lat", "age (Ma)"]

# Coregister deposit locations with cumulative erosion grids
# This links each deposit to its erosion history
if os.path.isfile(deposit_ero_filename):
    deposit_ero = pd.read_csv(deposit_ero_filename)
else:
    deposit_data_filtered = pd.read_csv(deposit_data_filtered_filename)
    deposit_ero = deposit_data_filtered[columns_ero]
    
    deposit_ero = run_coregister_erosion(
        point_data=deposit_ero,
        input_dir=ero_dir,
        distance_threshold=0.1,
        output_filename=deposit_ero_filename,
        n_jobs=n_jobs,
        verbose=True,
    )

# Coregister unlabelled samples with cumulative erosion grids
if os.path.isfile(unlabelled_ero_filename):
    unlabelled_ero = pd.read_csv(unlabelled_ero_filename)
else:
    unlabelled_data_filtered = pd.read_csv(unlabelled_data_filtered_filename)
    unlabelled_ero = unlabelled_data_filtered[columns_ero]
    
    unlabelled_ero = run_coregister_erosion(
        point_data=unlabelled_ero,
        input_dir=ero_dir,
        distance_threshold=0.1,
        output_filename=unlabelled_ero_filename,
        n_jobs=n_jobs,
        verbose=True,
    )

# Coregister target points generated within arc-backarc environments with cumulative erosion grids
if os.path.isfile(backarc_ero_filename):
    backarc_ero = pd.read_csv(backarc_ero_filename)
else:
    backarc_prob = pd.read_csv(backarc_prob_filename)
    backarc_ero = backarc_prob[columns_ero]
    
    backarc_ero = run_coregister_erosion(
        point_data=backarc_ero,
        input_dir=ero_dir,
        distance_threshold=0.1,
        output_filename=backarc_ero_filename,
        n_jobs=n_jobs,
        verbose=True,
    )

# Coregister target points within continental landmasses with cumulative erosion grids
if os.path.isfile(coastline_ero_filename):
    coastline_ero = pd.read_csv(coastline_ero_filename)
else:
    coastline_prob = pd.read_csv(coastline_prob_filename)
    columns_ero.append("index")
    coastline_ero = coastline_prob[columns_ero]
    
    coastline_ero = run_coregister_erosion(
        point_data=coastline_ero,
        input_dir=ero_dir,
        distance_threshold=0.1,
        output_filename=coastline_ero_filename,
        n_jobs=n_jobs,
        verbose=True,
    )
```

Here, we generate a plot comparing the distribution of cumulative erosion values between known mineral occurrences and unlabelled samples.

```python
deposit_ero_col = deposit_ero["erosion (m)"]
unlabelled_ero_col = unlabelled_ero["erosion (m)"]

# Compute quartiles
Q1 = np.percentile(deposit_ero_col, 25)
Q3 = np.percentile(deposit_ero_col, 75)
IQR = Q3 - Q1
# Upper threshold for outliers
upper_threshold = Q3 + 1.5 * IQR
print("Upper outlier threshold for positive samples:", upper_threshold)

# Fontsize controls
label_fontsize = 14
tick_fontsize = 12
legend_fontsize = 12
annotation_fontsize = 12

fig, ax = plt.subplots(figsize=(12, 6))
ax.set_facecolor("whitesmoke")
ax.scatter(unlabelled_ero["age (Ma)"], unlabelled_ero_col, facecolor="LightSalmon", edgecolor="black",
           s=10, alpha=0.2, label="Negative")
ax.scatter(deposit_ero["age (Ma)"], deposit_ero_col, facecolor="DarkSeaGreen", edgecolor="black",
           s=25, alpha=0.8, label="Positive")

# X axis: time_max on the left, time_min on the right
ax.set_xlim(time_max, time_min)
# Y axis: 0 at the top, increasing erosion downward
y_max = max(unlabelled_ero_col.max(), deposit_ero_col.max())
ax.set_ylim(y_max, 0)
# Horizontal threshold line
ax.axhline(upper_threshold, color="black", linestyle="--", linewidth=1.2)
# Annotate on the right side with a small arrow pointing to the line
ax.annotate(
    "Upper outlier threshold\nfor positive samples",
    xy=(1.0, upper_threshold), xycoords=("axes fraction", "data"),
    xytext=(1.05, upper_threshold), textcoords=("axes fraction", "data"),
    ha="left", va="center", fontsize=annotation_fontsize,
    arrowprops=dict(arrowstyle="->", color="black", lw=1),
)
ax.set_xlabel("Age (Ma)", fontsize=label_fontsize)
ax.set_ylabel("Cumulative erosion (m)", fontsize=label_fontsize)
ax.tick_params(axis='both', labelsize=tick_fontsize)
ax.grid(True, linestyle="--", alpha=0.5)
ax.legend(loc="best", fontsize=legend_fontsize)
plt.tight_layout()

fig_path = os.path.join(figs_dir, "cumulative_erosion_vs_age.png")
# if not os.path.exists(fig_path):
#     fig.savefig(fig_path, dpi=300, bbox_inches="tight")
    
plt.show()
```

```python
if os.path.isfile(pres_score_filename):
    pres_score = joblib.load(pres_score_filename)
else:
    # Exclude zero cumulative erosion. It is identically zero at 0 Ma by
    # construction and so carries no information about preservation.
    deposit_fit = deposit_ero_col.values[deposit_ero_col.values > 0]
    background_fit = unlabelled_ero_col.values[unlabelled_ero_col.values > 0]

    pres_score = PreservationScore(n_knots=6).fit(deposit_fit, background_fit)

    auc_mean, auc_sd = pres_score.held_out_auc(deposit_fit, background_fit)
    print(f"Held-out AUC = {auc_mean:.3f} +/- {auc_sd:.3f}")
    print(pres_score.summary().to_string(index=False))

    dump(pres_score, pres_score_filename, compress=True)

    print(f"Writing preservation score model to file: {pres_score_filename}")
```

### Preservation Score for Arc-Backarc Points

We estimate the preservation score for grid points generated within buffer zones covering arc–backarc environments by scoring their cumulative erosion with the case-control model fitted above. Each point is passed through the fitted spline response and returned on a 0–1 scale, using the bounds stored on the fitted object so that every point set scored with it shares one common scale and remains spatially comparable. The result highlights areas where mineral deposits are more likely to have survived erosion and to sit at a depth where they could be detected. These values are then stored in the coregistered point file and used to generate gridded spatiotemporal maps, supporting mineral prospectivity analysis.

```python
if "preservation_score" in backarc_ero.columns:
    print(f"Preservation scores have already been calculated and saved in {backarc_ero_filename}")
else:
    backarc_ero["preservation_score"] = pres_score.score(
        backarc_ero["erosion (m)"].values
    )
    backarc_ero.to_csv(backarc_ero_filename, index=False)

# Generate time-resolved preservation grids for the entire arc-backarc environments
if os.path.exists(backarc_pres_grids_dir):
    print(f"Preservation grids are located in {backarc_pres_grids_dir}")
else:
    os.makedirs(backarc_pres_grids_dir, exist_ok=True)

    create_grids(
        backarc_ero,
        output_dir=backarc_pres_grids_dir,
        times=time_steps,
        resolution=grid_resolution,
        extent=(-180, 180, -90, 90),
        threads=n_jobs,
        verbose=True,
        column="preservation_score",
        filename_format="backarc_preservation_grid_{}Ma.nc",
    )
```

### Preservation Maps for Arc-Backarc Points

The cell below loads the necessary files to create the `PlateReconstruction` and `PlateTopologies` objects, which will later be used for reconstruction and generating visualisations.

```python
# Muller25 plate motion model
pmm = PlateModelManager()
pm = pmm.get_model(plate_model)

rotation_model = pm.get_rotation_model()
topology_features = pm.get_topologies()

static_polygons = pm.get_static_polygons()
coastlines = pm.get_coastlines()
continents = pm.get_continental_polygons()
COBs = pm.get_COBs()

plate_reconstruction = PlateReconstruction(
    rotation_model=rotation_model,
    topology_features=topology_features,
    static_polygons=static_polygons,
    anchor_plate_id=anchor_plate_id,
)

gplot = PlotTopologies(
    plate_reconstruction=plate_reconstruction,
    coastlines=coastlines,
    continents=continents,
    COBs=COBs,
    anchor_plate_id=anchor_plate_id,
)

# Define map projection (Mollweide provides a good global view) and central meridian
projection = ccrs.Mollweide(central_longitude=30)
central_meridian = 0
```

This section generates an interactive preservation score map that displays values along trench lines, with the option to overlay known mineral occurrences. It also shows seafloor age and plate boundaries—including mid-ocean ridges and transform faults—at a selected time. The visualisation highlights areas favourable for the preservation of mineralisation, as well as regions subject to higher erosion rates that may have removed deposits or transported them to adjacent areas.

```python
@interact
def show_map(time=time_steps):
    gplot.time = time # Set reconstruction time
    fig = plot(
        gplot=gplot,
        grid=os.path.join(
            backarc_pres_grids_dir,
            f"backarc_preservation_grid_{time:0.0f}Ma.nc", # Preservation score grid
        ),
        grid_type="preservation",
        background_grid=os.path.join(
            agegrid_dir,
            f"seafloor_age_{time:0.0f}Ma.nc", # Seafloor age grid
        ),
        # positives=deposit_coords_recon_all_filtered_filename, # Reconstructed location of known mineral occurrences
        positives=None,
        projection=projection,
        time=time,
        central_meridian=central_meridian,
    )
```

The cell below generates a full sequence of preservation score maps through geological time and compiles them into a time-lapse animation, illustrating how the modelled preservation score changes over time.

```python
# Generate all preservation score maps and create animation
if os.path.exists(backarc_pres_maps_dir):
    print(f"Preservation score maps are located in {backarc_pres_maps_dir}")
else:
    os.makedirs(backarc_pres_maps_dir, exist_ok=True)

    # Create maps in parallel for efficiency
    with Parallel(
        n_jobs=n_jobs,
        verbose=True,
        pre_dispatch="all",
        batch_size=int(len(time_steps) // n_jobs) + 1,
    ) as parallel:

        # Define output filename for each time step
        output_filenames = [
            os.path.join(backarc_pres_maps_dir, f"backarc_preservation_map_{t:0.0f}Ma.png")
            for t in time_steps
        ]

        parallel(
            delayed(plot_parallel)(
                rotation_model,
                topology_features,
                static_polygons,
                coastlines,
                continents,
                COBs,
                anchor_plate_id,
                grid=os.path.join(
                    backarc_pres_grids_dir,
                    f"backarc_preservation_grid_{t:0.0f}Ma.nc",
                ),
                grid_type="preservation",
                background_grid=os.path.join(
                    agegrid_dir,
                    f"seafloor_age_{t:0.0f}Ma.nc",
                ),
                # positives=deposit_recon_coords_all_filtered_filename,
                positives=None,
                projection=projection,
                time=t,
                output_filename=o,
                central_meridian=central_meridian,
            )
            for t, o in zip(time_steps, output_filenames)
        )

    # Create an animation showing the evolution of preservation score
    output_filename = os.path.join(pres_dir, "backarc_preservation_animation.mp4")
    create_animation(
        image_filenames=output_filenames[::-1],
        output_filename=output_filename,
        fps=10,
        bitrate="5000k",
    )
```

### Preserved Mineralisation Prospectivity for Arc-Backarc Points

Here we combine mineralisation probability with preservation score to estimate the probability that mineral deposits are both formed and retained in arc–backarc environments through time. Mineralisation probability is multiplied by the scaled preservation score, and the combined results are stored as time-dependent grids. These grids display the spatial distribution of preserved mineralisation prospectivity, accounting for erosion and exhumation. This approach enables more realistic mineral prospectivity mapping by explicitly incorporating preservation bias.

### Preservation Score for Points within Continental Landmasses

Similar to arc-backarc points, we calculate the preservation score for grid points generated within continental landmasses by scoring their cumulative erosion with the same fitted case-control model. Each point is passed through the fitted spline response and returned on a 0–1 scale. Because the same fitted object is reused here, the bounds are those established at fitting time, so the continental and arc–backarc scores share a single common scale and can be compared directly. The result highlights areas where mineral deposits are more likely to have survived erosion and to sit at a depth where they could be detected. These values are stored and used to generate gridded spatiotemporal maps, as well as a single global mineral prospectivity map in present-day coordinates. The global map integrates the history of all points within continental landmasses, accounting for both mineralisation and preservation.

### Trend of Average Prospectivity and Adjusted Prospectivity

Here, we calculate the average preserved mineralisation prospectivity values greater than 0.5 for each of the points generated at the beginning of this notebook over time. By filtering out low-prospectivity areas (prospectivity ≤ 0.5), we focus on promising sites, resulting in a clean dataset that highlights the most probable and geologically reliable locations for porphyry mineralisation and preservation at present-day coordinates.

```python
coastline_data = pd.read_csv(coastline_data_filename)
mask = build_filter_mask(coastline_data, buffer_distance=buffer_distance)
coastline_data_filtered = filter_data(coastline_data, buffer_distance=buffer_distance)
coastline_prob_filtered = coastline_prob[mask.to_numpy()].reset_index(drop=True)

coastline_prob_present_day = pd.read_csv(coastline_prob_present_day_filename)
coastline_prob_unc_present_day = pd.read_csv(coastline_prob_unc_present_day_filename)

if "probability_pres" in coastline_prob_present_day.columns:
    print(f"Preserved mineralisation prospectivity has already been calculated and saved in {coastline_prob_present_day_filename}")
else:
    # Filter for high-prospectivity locations
    coastline_prob_high = coastline_prob_filtered[coastline_prob_filtered["probability_pres"] > 0.5]
    # Average preserved mineralisation prospectivity for each unique location index
    coastline_prob_high_mean = coastline_prob_high.groupby("index")[["probability_pres"]].mean().reset_index()
    # Merge the averaged prospectivities with present-day coordinates
    coastline_prob_present_day = pd.merge(coastline_prob_present_day, coastline_prob_high_mean, on="index", how="left")
    coastline_prob_present_day = coastline_prob_present_day[["index", "present_lon", "present_lat", "probability", "entropy", "vote_variance", "uncertainty", "probability_uncertainty", "probability_pres"]]
    # Save output
    coastline_prob_present_day.to_csv(coastline_prob_present_day_filename, index=False)

if "probability_unc_pres" in coastline_prob_unc_present_day.columns:
    print(f"Preserved mineralisation prospectivity has already been calculated and saved in {coastline_prob_unc_present_day_filename}")
else:
    # Filter for high-prospectivity locations
    coastline_prob_unc_high = coastline_prob_filtered[coastline_prob_filtered["probability_unc_pres"] > 0.5]
    # Average adjusted preserved mineralisation prospectivity for each unique location index
    coastline_prob_unc_high_mean = coastline_prob_unc_high.groupby("index")[["probability_unc_pres"]].mean().reset_index()
    # Merge the averaged prospectivities with present-day coordinates
    coastline_prob_unc_present_day = pd.merge(coastline_prob_unc_present_day, coastline_prob_unc_high_mean, on="index", how="left")
    coastline_prob_unc_present_day = coastline_prob_unc_present_day[["index", "present_lon", "present_lat", "probability", "entropy", "vote_variance", "uncertainty", "probability_uncertainty", "probability_unc_pres"]]
    # Save output
    coastline_prob_unc_present_day.to_csv(coastline_prob_unc_present_day_filename, index=False)
```

This section visualises how the average preservation score and prospectivity of preserved mineralisation for points within the continental landmasses vary over geological time, and relates these probabilities to both the number of known deposits at different ages (shown as a histogram) and the average value of a selected feature associated with those prospectivities. This highlights how the feature evolves through time.

It also provides the option to plot the uncertainty envelope for the average trends by calculating the standard deviation or median absolute deviation of values at each time step. The curves are smoothed to emphasise trends, and dual y-axes enable direct comparison between prospectivities, deposit frequency, and feature values, providing a comprehensive temporal overview. In doing so, the visualisation reveals how the preservation score and prospectivity of preserved mineralisation change across Earth's history, helping to identify key time periods favourable for porphyry formation and preservation.

```python
model_rf_calibrated = joblib.load(model_rf_calibrated_filename)
selected_features = model_rf_calibrated.feature_names_in_

deposit_coords_recon = pd.read_csv(deposit_coords_recon_filename)

# Prospectivity stats
coastline_prob_pres_mean_age = (
    coastline_prob.groupby("age (Ma)")["probability_pres"]
    .mean()
    .reset_index()
)

coastline_prob_unc_pres_mean_age = (
    coastline_prob.groupby("age (Ma)")["probability_unc_pres"]
    .mean()
    .reset_index()
)

coastline_pres_mean_age = (
    coastline_prob.groupby("age (Ma)")["preservation_score"]
    .mean()
    .reset_index()
)

coastline_prob_pres_mean_age = coastline_prob_pres_mean_age.sort_values("age (Ma)")
coastline_prob_unc_pres_mean_age = coastline_prob_unc_pres_mean_age.sort_values("age (Ma)")
coastline_pres_mean_age = coastline_pres_mean_age.sort_values("age (Ma)")

ages = coastline_prob_pres_mean_age["age (Ma)"].values.reshape(-1, 1)
probs_pres = coastline_prob_pres_mean_age["probability_pres"].values.reshape(-1, 1)
probs_unc_pres = coastline_prob_unc_pres_mean_age["probability_unc_pres"].values.reshape(-1, 1)
pres = coastline_pres_mean_age["preservation_score"].values.reshape(-1, 1)

# Smooth mean once (does not depend on spread type)
ages_smooth, probs_pres_smooth = smooth_curve(
    ages, probs_pres, method="gaussian", fwhm_myr=25
)

_, probs_unc_pres_smooth = smooth_curve(
    ages, probs_unc_pres, method="gaussian", fwhm_myr=25
)

_, pres_smooth = smooth_curve(
    ages, pres, method="gaussian", fwhm_myr=25
)

bin_width = 10
bins = np.arange(time_min, time_max + bin_width, bin_width)

# Interactive plot
@interact
def show_map(feature=selected_features, spread_method=["std", "mad"]):
    # Prospectivity spread
    coastline_prob_pres_spread_age = (
        coastline_prob.groupby("age (Ma)")["probability_pres"]
        .apply(lambda x: compute_spread(x, method=spread_method))
        .reset_index(name="probability_pres")
        .sort_values("age (Ma)")
    )

    coastline_prob_unc_pres_spread_age = (
        coastline_prob.groupby("age (Ma)")["probability_unc_pres"]
        .apply(lambda x: compute_spread(x, method=spread_method))
        .reset_index(name="probability_unc_pres")
        .sort_values("age (Ma)")
    )

    coastline_pres_spread_age = (
        coastline_prob.groupby("age (Ma)")["preservation_score"]
        .apply(lambda x: compute_spread(x, method=spread_method))
        .reset_index(name="preservation_score")
        .sort_values("age (Ma)")
    )

    probs_pres_spread = coastline_prob_pres_spread_age["probability_pres"].values.reshape(-1, 1)
    probs_unc_pres_spread = coastline_prob_unc_pres_spread_age["probability_unc_pres"].values.reshape(-1, 1)
    pres_spread = coastline_pres_spread_age["preservation_score"].values.reshape(-1, 1)

    _, probs_pres_spread_smooth = smooth_curve(
        ages, probs_pres_spread, method="gaussian", fwhm_myr=25
    )

    _, probs_unc_pres_spread_smooth = smooth_curve(
        ages, probs_unc_pres_spread, method="gaussian", fwhm_myr=25
    )

    _, pres_spread_smooth = smooth_curve(
        ages, pres_spread, method="gaussian", fwhm_myr=25
    )

    # Feature mean + spread
    coastline_feat_mean_age = (
        coastline_data_filtered.groupby("age (Ma)")[feature]
        .mean()
        .reset_index()
        .sort_values("age (Ma)")
    )

    coastline_feat_spread_age = (
        coastline_data_filtered.groupby("age (Ma)")[feature]
        .apply(lambda x: compute_spread(x, method=spread_method))
        .reset_index(name=feature)
        .sort_values("age (Ma)")
    )

    feats = coastline_feat_mean_age[feature].values.reshape(-1, 1)
    feats_spread = coastline_feat_spread_age[feature].values.reshape(-1, 1)

    _, feats_smooth = smooth_curve(
        ages, feats, method="gaussian", fwhm_myr=25
    )

    _, feats_spread_smooth = smooth_curve(
        ages, feats_spread, method="gaussian", fwhm_myr=25
    )

    # Plot
    fig, ax1 = plt.subplots(figsize=(10, 6))
    ax1.set_facecolor("whitesmoke")

    # Histogram
    ax1.hist(
        deposit_coords_recon["age (Ma)"],
        bins=bins,
        color="gray",
        alpha=0.5,
        edgecolor="black",
        zorder=1,
    )
    ax1.set_xlabel("Age (Ma)", fontsize=14)
    ax1.set_ylabel("Deposit count", fontsize=14)

    # Prospectivity axis
    ax2 = ax1.twinx()

    ax2.plot(
        ages_smooth,
        probs_pres_smooth,
        color="orangered",
        linewidth=2,
        label="Preserved mineralisation prospectivity",
        zorder=2,
    )

    ax2.plot(
        ages_smooth,
        probs_unc_pres_smooth,
        color="forestgreen",
        linewidth=2,
        label="Adjusted probability",
        zorder=2,
    )

    ax2.plot(
        ages_smooth,
        pres_smooth,
        color="magenta",
        linewidth=2,
        label="Preservation score",
        zorder=2,
    )

    ax2.fill_between(
        ages_smooth.flatten(),
        (probs_pres_smooth - probs_pres_spread_smooth).flatten(),
        (probs_pres_smooth + probs_pres_spread_smooth).flatten(),
        color="orangered",
        alpha=0.2,
    )

    ax2.fill_between(
        ages_smooth.flatten(),
        (probs_unc_pres_smooth - probs_unc_pres_spread_smooth).flatten(),
        (probs_unc_pres_smooth + probs_unc_pres_spread_smooth).flatten(),
        color="forestgreen",
        alpha=0.2,
    )

    ax2.fill_between(
        ages_smooth.flatten(),
        (pres_smooth - pres_spread_smooth).flatten(),
        (pres_smooth + pres_spread_smooth).flatten(),
        color="magenta",
        alpha=0.2,
    )

    ax2.set_xlim(time_max, time_min)
    ax2.set_ylim(0, 1)
    ax2.set_ylabel("Mineralisation prospectivity / preservation score", fontsize=14)

    # Feature axis (3rd axis)
    feature_label = format_feature_name(feature)
    uncertainty_label = "±1σ" if spread_method == "std" else "±MAD"
    feature_label_with_uncertainty = f"{feature_label} ({uncertainty_label})"

    ax3 = ax1.twinx()
    ax3.spines["right"].set_position(("outward", 60))

    ax3.plot(
        ages_smooth,
        feats_smooth,
        color="royalblue",
        linewidth=2,
        label=feature_label_with_uncertainty,
        zorder=2,
    )

    ax3.fill_between(
        ages_smooth.flatten(),
        (feats_smooth - feats_spread_smooth).flatten(),
        (feats_smooth + feats_spread_smooth).flatten(),
        color="royalblue",
        alpha=0.2,
    )

    ax3.set_ylabel(feature_label, fontsize=14)

    ax1.tick_params(labelsize=12)
    ax2.tick_params(labelsize=12)
    ax3.tick_params(labelsize=12)

    # Legend
    spread_label = "±1σ" if spread_method == "std" else "±MAD"

    lines2, labels2 = ax2.get_legend_handles_labels()
    lines3, labels3 = ax3.get_legend_handles_labels()

    labels2 = [f"{l} ({spread_label})" for l in labels2]

    lines = lines2 + lines3
    labels = labels2 + labels3

    ax3.legend(lines, labels, loc="upper left", frameon=True, fontsize=12)

    plt.tight_layout()
    
    clean_feature = clean_feature_name(feature)
    fig_path = os.path.join(figs_dir, f"coastline_prob_pres_avg_trend_{clean_feature}.png")
    # if not os.path.exists(fig_path):
    #     plt.savefig(
    #         fig_path,
    #         dpi=300,
    #         bbox_inches="tight",
    #     )
    
    plt.show()
```

```python
# Get ages with prospectivity > 0.5
coastline_prob_pres_high = coastline_prob[coastline_prob["probability_pres"] > 0.5]
coastline_prob_unc_pres_high = coastline_prob[coastline_prob["probability_unc_pres"] > 0.5]

coastline_prob_pres_high_mean_age = (
    coastline_prob_pres_high.groupby("age (Ma)")["probability_pres"]
    .mean()
    .reset_index()
)

coastline_pres_high_mean_age = (
    coastline_prob_pres_high.groupby("age (Ma)")["preservation_score"]
    .mean()
    .reset_index()
)

coastline_prob_unc_pres_high_mean_age = (
    coastline_prob_unc_pres_high.groupby("age (Ma)")["probability_unc_pres"]
    .mean()
    .reset_index()
)

coastline_pres_adj_high_mean_age = (
    coastline_prob_unc_pres_high.groupby("age (Ma)")["preservation_score"]
    .mean()
    .reset_index()
)

# Get all unique ages
all_ages = coastline_prob["age (Ma)"].unique()
complete_ages = pd.DataFrame({"age (Ma)": all_ages})

# Fill missing
coastline_prob_pres_high_mean_age = complete_ages.merge(
    coastline_prob_pres_high_mean_age,
    on="age (Ma)",
    how="left"
).fillna(coastline_prob_pres_high_mean_age["probability_pres"].mean())

coastline_pres_high_mean_age = complete_ages.merge(
    coastline_pres_high_mean_age,
    on="age (Ma)",
    how="left"
).fillna(coastline_pres_high_mean_age["preservation_score"].mean())

coastline_prob_unc_pres_high_mean_age = complete_ages.merge(
    coastline_prob_unc_pres_high_mean_age,
    on="age (Ma)",
    how="left"
).fillna(coastline_prob_unc_pres_high_mean_age["probability_unc_pres"].mean())

coastline_pres_adj_high_mean_age = complete_ages.merge(
    coastline_pres_adj_high_mean_age,
    on="age (Ma)",
    how="left"
).fillna(coastline_pres_adj_high_mean_age["preservation_score"].mean())

coastline_prob_pres_high_mean_age = coastline_prob_pres_high_mean_age.sort_values("age (Ma)")
coastline_pres_high_mean_age = coastline_pres_high_mean_age.sort_values("age (Ma)")
coastline_prob_unc_pres_high_mean_age = coastline_prob_unc_pres_high_mean_age.sort_values("age (Ma)")
coastline_pres_adj_high_mean_age = coastline_pres_adj_high_mean_age.sort_values("age (Ma)")

ages_high = coastline_prob_pres_high_mean_age["age (Ma)"].values.reshape(-1, 1)
probs_pres_high = coastline_prob_pres_high_mean_age["probability_pres"].values.reshape(-1, 1)
pres_high = coastline_pres_high_mean_age["preservation_score"].values.reshape(-1, 1)
probs_unc_pres_high = coastline_prob_unc_pres_high_mean_age["probability_unc_pres"].values.reshape(-1, 1)
pres_adj_high = coastline_pres_adj_high_mean_age["preservation_score"].values.reshape(-1, 1)

# Smooth mean once
ages_high_smooth, probs_pres_high_smooth = smooth_curve(
    ages_high, probs_pres_high, method="gaussian", fwhm_myr=25
)

_, pres_high_smooth = smooth_curve(
    ages_high, pres_high, method="gaussian", fwhm_myr=25
)

_, probs_unc_pres_high_smooth = smooth_curve(
    ages_high, probs_unc_pres_high, method="gaussian", fwhm_myr=25
)

_, pres_adj_high_smooth = smooth_curve(
    ages_high, pres_adj_high, method="gaussian", fwhm_myr=25
)

# Interactive plot
@interact
def show_map(
    feature=selected_features,
    spread_method=["std", "mad"],
    probability_type=[
        "Mineralisation prospectivity",
        "Adjusted prospectivity"
    ]
):
    if probability_type == "Mineralisation prospectivity":
        prob_df = coastline_prob_pres_high
        prob_col = "probability_pres"
        prob_mean_smooth = probs_pres_high_smooth
        pres_mean_smooth = pres_high_smooth
        prob_color = "orangered"
        prob_label = "Mineralisation prospectivity"
        pres_label = "Preservation score"
    else:
        prob_df = coastline_prob_unc_pres_high
        prob_col = "probability_unc_pres"
        prob_mean_smooth = probs_unc_pres_high_smooth
        pres_mean_smooth = pres_adj_high_smooth
        prob_color = "forestgreen"
        prob_label = "Adjusted prospectivity"
        pres_label = "Adjusted preservation score"
        
    # Prospectivity spread
    prob_spread_age = (
        prob_df.groupby("age (Ma)")[prob_col]
        .apply(lambda x: compute_spread(x, method=spread_method))
        .reset_index(name=prob_col)
    )

    prob_spread_age = complete_ages.merge(
        prob_spread_age,
        on="age (Ma)",
        how="left"
    ).fillna(prob_spread_age[prob_col].mean())

    prob_spread_age = prob_spread_age.sort_values("age (Ma)")

    prob_spread = prob_spread_age[prob_col].values.reshape(-1, 1)

    _, prob_spread_smooth = smooth_curve(
        ages_high, prob_spread, method="gaussian", fwhm_myr=25
    )

    # Preservation score spread for selected prospectivity type
    pres_spread_age = (
        prob_df.groupby("age (Ma)")["preservation_score"]
        .apply(lambda x: compute_spread(x, method=spread_method))
        .reset_index(name="preservation_score")
    )

    pres_spread_age = complete_ages.merge(
        pres_spread_age,
        on="age (Ma)",
        how="left"
    ).fillna(pres_spread_age["preservation_score"].mean())

    pres_spread_age = pres_spread_age.sort_values("age (Ma)")
    pres_spread = pres_spread_age["preservation_score"].values.reshape(-1, 1)

    _, pres_spread_smooth = smooth_curve(
        ages_high, pres_spread, method="gaussian", fwhm_myr=25
    )

    # Feature mean + spread for selected prospectivity type
    coastline_feat_high = prob_df.merge(
        coastline_data_filtered,
        left_index=True,
        right_index=True,
        suffixes=('', '_drop')
    ).filter(regex='^(?!.*_drop)')

    # Feature mean
    coastline_feat_high_mean_age = (
        coastline_feat_high.groupby("age (Ma)")[feature]
        .mean()
        .reset_index()
    )

    coastline_feat_high_mean_age = complete_ages.merge(
        coastline_feat_high_mean_age,
        on="age (Ma)",
        how="left"
    ).fillna(coastline_feat_high_mean_age[feature].mean())

    # Feature spread
    coastline_feat_high_spread_age = (
        coastline_feat_high.groupby("age (Ma)")[feature]
        .apply(lambda x: compute_spread(x, method=spread_method))
        .reset_index(name=feature)
    )

    coastline_feat_high_spread_age = complete_ages.merge(
        coastline_feat_high_spread_age,
        on="age (Ma)",
        how="left"
    ).fillna(coastline_feat_high_spread_age[feature].mean())

    # Sort
    coastline_feat_high_mean_age = coastline_feat_high_mean_age.sort_values("age (Ma)")
    coastline_feat_high_spread_age = coastline_feat_high_spread_age.sort_values("age (Ma)")

    # Arrays
    feat_vals = coastline_feat_high_mean_age[feature].values.reshape(-1, 1)
    feat_spread = coastline_feat_high_spread_age[feature].values.reshape(-1, 1)

    # Smooth
    _, feat_vals_smooth = smooth_curve(
        ages_high, feat_vals, method="gaussian", fwhm_myr=25
    )

    _, feat_spread_smooth = smooth_curve(
        ages_high, feat_spread, method="gaussian", fwhm_myr=25
    )

    # Plot
    fig, ax1 = plt.subplots(figsize=(10, 6))
    ax1.set_facecolor("whitesmoke")

    ax1.hist(
        deposit_coords_recon["age (Ma)"],
        bins=bins,
        color="gray",
        alpha=0.5,
        edgecolor="black"
    )
    ax1.set_xlabel("Age (Ma)", fontsize=14)
    ax1.set_ylabel("Deposit count", fontsize=14)

    # Prospectivity axis
    ax2 = ax1.twinx()

    ax2.plot(
        ages_high_smooth,
        prob_mean_smooth,
        color=prob_color,
        linewidth=2,
        label=prob_label,
    )

    ax2.fill_between(
        ages_high_smooth.flatten(),
        (prob_mean_smooth - prob_spread_smooth).flatten(),
        (prob_mean_smooth + prob_spread_smooth).flatten(),
        color=prob_color,
        alpha=0.2,
    )

    ax2.plot(
        ages_high_smooth,
        pres_mean_smooth,
        color="magenta",
        linewidth=2,
        label=pres_label,
    )

    ax2.fill_between(
        ages_high_smooth.flatten(),
        (pres_mean_smooth - pres_spread_smooth).flatten(),
        (pres_mean_smooth + pres_spread_smooth).flatten(),
        color="magenta",
        alpha=0.2,
    )

    ax2.set_xlim(time_max, time_min)
    ax2.set_ylim(0.5, 1)
    ax2.set_ylabel(f"{prob_label} / preservation score (P > 0.5)", fontsize=12)

    # Feature axis
    feature_label = format_feature_name(feature)
    uncertainty_label = "±1σ" if spread_method == "std" else "±MAD"
    feature_label_with_uncertainty = f"{feature_label} ({uncertainty_label})"

    ax3 = ax1.twinx()
    ax3.spines["right"].set_position(("outward", 60))

    ax3.plot(
        ages_high_smooth,
        feat_vals_smooth,
        color="royalblue",
        linewidth=2,
        label=feature_label_with_uncertainty
    )

    ax3.fill_between(
        ages_high_smooth.flatten(),
        (feat_vals_smooth - feat_spread_smooth).flatten(),
        (feat_vals_smooth + feat_spread_smooth).flatten(),
        color="royalblue",
        alpha=0.2,
    )

    ax3.set_ylabel(feature_label, fontsize=14)

    ax1.tick_params(labelsize=12)
    ax2.tick_params(labelsize=12)
    ax3.tick_params(labelsize=12)

    # Legend
    spread_label = "±1σ" if spread_method == "std" else "±MAD"

    lines2, labels2 = ax2.get_legend_handles_labels()
    lines3, labels3 = ax3.get_legend_handles_labels()

    labels2 = [f"{l} ({spread_label})" for l in labels2]

    lines = lines2 + lines3
    labels = labels2 + labels3

    ax3.legend(lines, labels, loc="upper left", frameon=True, fontsize=12)

    plt.tight_layout()

    clean_feature = clean_feature_name(feature)
    clean_prob_type = clean_feature_name(probability_type)
    fig_path = os.path.join(
        figs_dir,
        f"coastline_prob_pres_avg_trend_high_{clean_prob_type}_{clean_feature}.png"
    )

    # if not os.path.exists(fig_path):
    #     plt.savefig(
    #         fig_path,
    #         dpi=300,
    #         bbox_inches="tight",
    #     )

    plt.show()
```

### Global Prospectivity Map of Preserved Porphyry Mineralisation

This code block generates two global grids of preserved mineralisation prospectivity and adjusted prospectivity for points within the present-day continental landmasses by averaging their values through time. The grids are created by interpolating these point values, with the interpolation step filling gaps between measurements. We use nearest-neighbor interpolation but mask out grid cells located far from any known data points (based on grid resolution) to avoid misleading extrapolations. The final gridded dataset is saved as a compressed NetCDF file.

```python
if not os.path.isfile(coastline_prob_pres_present_day_grid_filename) or not os.path.isfile(coastline_prob_unc_pres_present_day_grid_filename):
    # Create prospectivity grid for the points within present-day continental landmasses
    xmin, xmax, ymin, ymax = -180, 180, -90, 90

    # Create the meshes
    grid_lons = np.arange(xmin, xmax + grid_resolution, grid_resolution)
    grid_lats = np.arange(ymin, ymax + grid_resolution, grid_resolution)
    grid_mlons, grid_mlats = np.meshgrid(grid_lons, grid_lats)

    for filename, data_lons, data_lats, data_values in [
        (
            coastline_prob_pres_present_day_grid_filename,
            coastline_prob_present_day["present_lon"],
            coastline_prob_present_day["present_lat"],
            coastline_prob_present_day["probability_pres"],
        ),
        (
            coastline_prob_unc_pres_present_day_grid_filename,
            coastline_prob_unc_present_day["present_lon"],
            coastline_prob_unc_present_day["present_lat"],
            coastline_prob_unc_present_day["probability_unc_pres"],
        ),
    ]:
        if not os.path.isfile(filename):
            # Interpolate using the nearest neighbour method
            points = np.column_stack((data_lons, data_lats))
            arr = griddata(points, data_values, (grid_mlons, grid_mlats), method="nearest")  # or "linear", "cubic"
        
            # Mask distant grid nodes
            tree = cKDTree(points)
            flat_grid_points = np.column_stack((grid_mlons.ravel(), grid_mlats.ravel()))
            distances, _ = tree.query(flat_grid_points, k=1)
        
            mask = distances > grid_resolution
            arr.ravel()[mask] = np.nan  # Mask out distant nodes        
        
            # Create the dataset
            dset = xr.Dataset(
                data_vars={
                    "z": (("lat", "lon"), arr),
                },
                coords={
                    "lon": grid_lons,
                    "lat": grid_lats,
                },
            )
        
            # Set projection info
            dset.rio.write_crs(4326, inplace=True)
        
            # Save as NetCDF
            dset.to_netcdf(
                filename,
                encoding={
                    "z": {
                        "zlib": True,
                        "dtype": "float32",
                    }
                },
            )
```

### Prediction-Area Plot

This code segment generates a Prediction–Area (P–A) plot, a valuable tool for assessing the effectiveness of the model in predicting porphyry preserved mineralisation. The P–A plot quantitatively compares the spatial distribution of predicted probabilities with the known locations of mineral occurrences.

For each known mineral occurrence, the corresponding prospectivity value is extracted from the raster grid by selecting the nearest grid cell to the occurrence's geographic coordinates. This provides insights into the model's precision and efficiency in targeting exploration areas.

The analysis considers a range of prospectivity thresholds from 0 to 1. For each threshold, two key percentages are computed:

- **Occurrence Percentage**: The proportion of known mineral occurrences with predicted probabilities greater than or equal to the threshold. This measures how many actual deposits are captured by a given prospectivity cutoff.
- **Area Percentage**: The proportion of the study area with predicted probabilities exceeding the same threshold. This represents how much geographic area would need to be explored if the threshold were applied as a filter.

The resulting curve illustrates model performance by showing how increasing the prospectivity cutoff reduces the exploration area but may also exclude some known occurrences.

```python
# Load prospectivity data
prospectivity_data = xr.open_dataset(coastline_prob_pres_present_day_grid_filename)
# prospectivity_data = xr.open_dataset(coastline_prob_unc_pres_present_day_grid_filename)
prospectivity_data = prospectivity_data["z"].squeeze()

# Extract prospectivity values at known mineral occurrence locations
occurrence_probabilities = []
for _, row in deposit_coords_recon.iterrows():
    lon, lat = row["present_lon"], row["present_lat"]
    # Extract prospectivity value at this point
    prob_value = prospectivity_data.sel(lon=lon, lat=lat, method="nearest").item()
    occurrence_probabilities.append(prob_value)

# Convert to numpy array for easier manipulation
occurrence_probabilities = np.array(occurrence_probabilities)
occurrence_probabilities = occurrence_probabilities[~np.isnan(occurrence_probabilities)]

# Calculate total study area and number of valid cells
total_area = np.count_nonzero(~np.isnan(prospectivity_data))
total_occurrences = len(occurrence_probabilities)

# Define the prospectivity thresholds
thresholds = np.linspace(0, 1, 100)

# Initialize lists to store results
occurrence_percentages = []
area_percentages = []

# Calculate percentages for each threshold
for threshold in thresholds:
    occurrence_percentage = np.sum(occurrence_probabilities >= threshold) / total_occurrences * 100
    area_percentage = np.sum(prospectivity_data >= threshold) / total_area * 100
    occurrence_percentages.append(occurrence_percentage)
    area_percentages.append(area_percentage)

# Convert lists to numpy arrays for plotting
occurrence_percentages = np.array(occurrence_percentages)
area_percentages = np.array(area_percentages)

# Set the font sizes
SMALL_SIZE = 12
MEDIUM_SIZE = 14
LARGE_SIZE = 16

# Create the P-A plot
fig, ax1 = plt.subplots(figsize=(10, 6))

ax1.set_facecolor("whitesmoke")

# Plot the occurrence percentages
ax1.plot(thresholds, occurrence_percentages, color="royalblue", linewidth=2)
ax1.set_xlabel("Prospectivity", fontsize=MEDIUM_SIZE)
ax1.set_ylabel("Percentage of known mineral occurrences", color="royalblue", fontsize=MEDIUM_SIZE)
ax1.set_ylim(0, 100)
ax1.set_xlim(0.5, 1)
ax1.grid(True)

# Set tick label sizes
ax1.tick_params(axis="both", labelsize=SMALL_SIZE)

# Create a second y-axis to plot the area percentages
ax2 = ax1.twinx()
ax2.plot(thresholds, area_percentages, color="orangered", linewidth=2)
ax2.set_ylabel("Percentage of study area", color="orangered", fontsize=MEDIUM_SIZE)
ax2.set_ylim(100, 0)  # Inverted y-axis
ax2.tick_params(axis="y", labelsize=SMALL_SIZE)

# prob_pres
intersection_x = 0.605
intersection_y1 = 57
intersection_y2 = 43

# prob_unc_pres
# intersection_x = 0.565
# intersection_y1 = 55
# intersection_y2 = 45

h_line = ax1.hlines(y=intersection_y1, xmin=0, xmax=1, color="g", linestyles=":")
v_line = ax1.vlines(x=intersection_x, ymin=0, ymax=100, color="g", linestyles=":")

# Add a text box to explain the intersection point
ax1.annotate(f"At prospectivity {intersection_x}:\n{intersection_y1}% of occurrences\n{intersection_y2}% of area",
             xy=(0.45, 0.6), xycoords="axes fraction",
             ha="center", va="top",
             fontsize=SMALL_SIZE,
             bbox=dict(boxstyle="round", fc="white", ec="gray", alpha=0.8))

# plt.title("Prediction-Area Plot")
plt.tight_layout()  # Adjust layout to prevent clipping of labels

# Adjust the plot margins to make room for annotations
plt.subplots_adjust(left=0.15, right=0.85)

# pa_probability_pres, pa_probability_unc_pres
fig_path = os.path.join(figs_dir, "pa_probability_pres.png")
if not os.path.exists(fig_path):
    fig.savefig(
        fig_path,
        dpi=300,
        bbox_inches="tight",
    )

plt.show()
```
